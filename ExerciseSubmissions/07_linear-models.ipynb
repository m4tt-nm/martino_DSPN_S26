```r
#INSERT CODE HERE
perfGM_model <- lm(Flanker_Unadj ~ FS_Tot_GM_Vol, data = d1)

summary(perfGM_model)
coef(perfGM_model)
confint(perfGM_model)
```

> *The mean value for beta0 is 90.26 while the mean value for beta1 is 3.110* **10^-5. Their confidence intervals are [*84.29, 96.22*]* and *[2.24 * 10^-5, 3.98 * 10^-5]* *respectively. Gray matter volume is also significantly associated with flanker task performance.*

## 4. Plotting (2 points)

Use `ggplot` to plot the `FS_Total_GM_Vol` variable (x axis) against the `Flanker_Unadj` variable (y axis), as well as the regression line with confidence intervals on the regrssion line. Qualitatively describe what you see.

```r
#INSERT CODE HERE
ggplot(perfGM_model, aes(x = FS_Tot_GM_Vol, y = Flanker_Unadj)) +
  geom_point() +
  geom_smooth() +
  theme_minimal() +
  labs(x = 'Total Gray Matter Volume', y = 'Flanker Task Performance')
```
